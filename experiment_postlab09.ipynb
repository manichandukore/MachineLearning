```python
# ============================================================
# EXPERIMENT 9: CNN AND RNN MODELS
# POST-LAB TASK
# ============================================================

# Install/Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    mean_squared_error,
    mean_absolute_error
)

print("TensorFlow Version:", tf.__version__)


# ============================================================
# PART A: CNN - IMAGE CLASSIFICATION
# Dataset: CIFAR-10
# ============================================================

print("\n================ CNN MODEL ================\n")

# ------------------------------------------------------------
# 1. Load CIFAR-10 Dataset
# ------------------------------------------------------------

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

print("Training images:", x_train.shape)
print("Testing images :", x_test.shape)


# ------------------------------------------------------------
# 2. Normalize Images
# ------------------------------------------------------------

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0


# ------------------------------------------------------------
# 3. Display Sample Images
# ------------------------------------------------------------

class_names = [
    "airplane",
    "automobile",
    "bird",
    "cat",
    "deer",
    "dog",
    "frog",
    "horse",
    "ship",
    "truck"
]

plt.figure(figsize=(10, 6))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i])
    plt.title(class_names[y_train[i][0]])
    plt.axis("off")

plt.suptitle("CIFAR-10 Sample Images")
plt.show()


# ------------------------------------------------------------
# 4. Build CNN Model
# ------------------------------------------------------------

cnn_model = tf.keras.Sequential([

    # Convolution Layer 1
    tf.keras.layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        input_shape=(32, 32, 3)
    ),

    # Pooling Layer 1
    tf.keras.layers.MaxPooling2D((2, 2)),

    # Convolution Layer 2
    tf.keras.layers.Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),

    # Pooling Layer 2
    tf.keras.layers.MaxPooling2D((2, 2)),

    # Convolution Layer 3
    tf.keras.layers.Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),

    # Flatten
    tf.keras.layers.Flatten(),

    # Fully Connected Layer
    tf.keras.layers.Dense(
        64,
        activation="relu"
    ),

    # Output Layer
    tf.keras.layers.Dense(
        10,
        activation="softmax"
    )
])


# ------------------------------------------------------------
# 5. Display CNN Architecture
# ------------------------------------------------------------

cnn_model.summary()


# ------------------------------------------------------------
# 6. Compile CNN
# ------------------------------------------------------------

cnn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# 7. Train CNN
# ------------------------------------------------------------

cnn_history = cnn_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2
)


# ------------------------------------------------------------
# 8. Evaluate CNN
# ------------------------------------------------------------

cnn_loss, cnn_accuracy = cnn_model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nCNN Test Loss:", cnn_loss)
print("CNN Test Accuracy:", cnn_accuracy)


# ============================================================
# CNN GRAPH 1: TRAINING AND VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    cnn_history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    cnn_history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("CNN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

plt.show()


# ============================================================
# CNN GRAPH 2: TRAINING AND VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    cnn_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    cnn_history.history["val_loss"],
    label="Validation Loss"
)

plt.title("CNN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 9. CNN Predictions
# ------------------------------------------------------------

cnn_predictions = cnn_model.predict(x_test)

cnn_predicted_classes = np.argmax(
    cnn_predictions,
    axis=1
)

cnn_actual_classes = y_test.flatten()


# ------------------------------------------------------------
# 10. CNN Classification Report
# ------------------------------------------------------------

print("\n================ CNN CLASSIFICATION REPORT ================\n")

print(
    classification_report(
        cnn_actual_classes,
        cnn_predicted_classes,
        target_names=class_names
    )
)


# ============================================================
# CNN GRAPH 3: CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    cnn_actual_classes,
    cnn_predicted_classes
)

plt.figure(figsize=(10, 8))

plt.imshow(cm)

plt.title("CNN Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.colorbar()

plt.xticks(
    range(10),
    class_names,
    rotation=45
)

plt.yticks(
    range(10),
    class_names
)

plt.show()


# ============================================================
# PART B: RNN - TIME SERIES PREDICTION
# Dataset: Airline Passengers
# ============================================================

print("\n================ RNN MODEL ================\n")


# ------------------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------------------

url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"

data = pd.read_csv(url)

print(data.head())
print("\nDataset Shape:", data.shape)


# ------------------------------------------------------------
# 2. Display Original Time Series
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    data["Passengers"]
)

plt.title("Airline Passenger Data")
plt.xlabel("Time")
plt.ylabel("Number of Passengers")

plt.grid()

plt.show()


# ------------------------------------------------------------
# 3. Extract Passenger Values
# ------------------------------------------------------------

values = data["Passengers"].values.astype("float32")


# ------------------------------------------------------------
# 4. Normalize Data
# ------------------------------------------------------------

minimum = values.min()
maximum = values.max()

values_normalized = (
    values - minimum
) / (
    maximum - minimum
)


# ------------------------------------------------------------
# 5. Create Sequences
# ------------------------------------------------------------

sequence_length = 12

X = []
y = []

for i in range(
    len(values_normalized) - sequence_length
):

    X.append(
        values_normalized[
            i:i + sequence_length
        ]
    )

    y.append(
        values_normalized[
            i + sequence_length
        ]
    )


X = np.array(X)
y = np.array(y)


# ------------------------------------------------------------
# 6. Reshape Data for RNN
# ------------------------------------------------------------

X = X.reshape(
    X.shape[0],
    X.shape[1],
    1
)

print("\nRNN Input Shape:", X.shape)
print("RNN Output Shape:", y.shape)


# ------------------------------------------------------------
# 7. Train-Test Split
# ------------------------------------------------------------

split = int(
    len(X) * 0.8
)

X_train = X[:split]
X_test = X[split:]

y_train = y[:split]
y_test = y[split:]

print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# ------------------------------------------------------------
# 8. Build RNN Model
# ------------------------------------------------------------

rnn_model = tf.keras.Sequential([

    # RNN Layer
    tf.keras.layers.SimpleRNN(
        50,
        activation="tanh",
        input_shape=(sequence_length, 1)
    ),

    # Output Layer
    tf.keras.layers.Dense(1)
])


# ------------------------------------------------------------
# 9. Display RNN Architecture
# ------------------------------------------------------------

rnn_model.summary()


# ------------------------------------------------------------
# 10. Compile RNN
# ------------------------------------------------------------

rnn_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)


# ------------------------------------------------------------
# 11. Train RNN
# ------------------------------------------------------------

rnn_history = rnn_model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=16,
    validation_data=(X_test, y_test),
    verbose=1
)


# ------------------------------------------------------------
# 12. Evaluate RNN
# ------------------------------------------------------------

rnn_loss, rnn_mae = rnn_model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("\nRNN Test Loss:", rnn_loss)
print("RNN Test MAE:", rnn_mae)


# ============================================================
# RNN GRAPH 1: TRAINING AND VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    rnn_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    rnn_history.history["val_loss"],
    label="Validation Loss"
)

plt.title("RNN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# RNN GRAPH 2: TRAINING AND VALIDATION MAE
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    rnn_history.history["mae"],
    label="Training MAE"
)

plt.plot(
    rnn_history.history["val_mae"],
    label="Validation MAE"
)

plt.title("RNN Training and Validation MAE")
plt.xlabel("Epoch")
plt.ylabel("MAE")

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 13. Make Predictions
# ------------------------------------------------------------

predictions = rnn_model.predict(
    X_test
)

predictions = predictions.flatten()


# ------------------------------------------------------------
# 14. Calculate MSE and MAE
# ------------------------------------------------------------

mse = mean_squared_error(
    y_test,
    predictions
)

mae = mean_absolute_error(
    y_test,
    predictions
)

print("\n================ RNN PERFORMANCE ================\n")

print("Mean Squared Error :", mse)
print("Mean Absolute Error:", mae)


# ============================================================
# RNN GRAPH 3: ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    y_test,
    label="Actual"
)

plt.plot(
    predictions,
    label="Predicted"
)

plt.title("RNN Actual vs Predicted Values 24eu02023")
plt.xlabel("Time")
plt.ylabel("Normalized Passenger Value")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# RNN GRAPH 4: ACTUAL VS PREDICTED - ORIGINAL SCALE
# ============================================================

actual_original = (
    y_test * (maximum - minimum)
) + minimum

predicted_original = (
    predictions * (maximum - minimum)
) + minimum


plt.figure(figsize=(10, 5))

plt.plot(
    actual_original,
    label="Actual"
)

plt.plot(
    predicted_original,
    label="Predicted"
)

plt.title(
    "RNN Actual vs Predicted Passenger Numbers 24eu02023"
)

plt.xlabel("Time")
plt.ylabel("Number of Passengers ")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# FINAL PERFORMANCE COMPARISON
# ============================================================

print("\n================================================")
print("FINAL MODEL PERFORMANCE")
print("================================================")

print(
    "CNN Test Accuracy:",
    cnn_accuracy
)

print(
    "RNN Test MSE:",
    mse
)

print(
    "RNN Test MAE:",
    mae
)
```
